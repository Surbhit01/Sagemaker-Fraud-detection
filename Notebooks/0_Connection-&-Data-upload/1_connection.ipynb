# Connecting to SageMaker from a Local Machine

Before using the `sagemaker`/`boto3` SDKs for anything else, confirm the local AWS
credentials (from `aws configure` or SSO) resolve to the account and identity you expect.
Every later module assumes this check has already passed.

## 1. Confirm local AWS credentials

```python
import boto3


sts = boto3.client("sts")
identity = sts.get_caller_identity()
print(f"Account: {identity['Account']}")
print(f"Identity ARN: {identity['Arn']}")
```
